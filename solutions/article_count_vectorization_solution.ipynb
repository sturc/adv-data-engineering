# Count N-Grams Vectorization

```python
import glob
import os
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

## Read in the Data Files

```python
# https://archive.ics.uci.edu/dataset/311/sentence+classification 
input_folder = "../data/labeled_articles/"
file_list=glob.glob("../data/labeled_articles/*.txt")

dfs = []
for filename in file_list:
    #print(f"Processing file: {filename}")
    df = pd.read_csv(filename, delimiter='\t', names=['label', 'text'], index_col=None, header=None, on_bad_lines='skip')
    df['file'] = os.path.basename(filename)
    if df['text'].isnull().any():
        #print(f"Missing values found in file: {filename}")
        df['text']=df["label"].str[5:]  # Fix for files with missing text values
        df['label']=df["label"].str[:4]
    dfs.append(df)

df = pd.concat(dfs, ignore_index=True)
print(df.count())
print(df[df["text"].isnull()]["file"].unique().tolist())
```

## Build N-Grams

```python
cv = CountVectorizer()   
single_word = cv.fit_transform(df['text']) 
print(single_word.shape)  
pd.DataFrame(single_word.todense(), columns=cv.get_feature_names_out())
# those are the unique words found across all articles
print(cv.vocabulary_.keys())  
```

## Limit the vocabulary

```python
cv = CountVectorizer(max_features=20)   
limited_word = cv.fit_transform(df['text']) 
print(limited_word.shape)  
pd.DataFrame(limited_word.todense(), columns=cv.get_feature_names_out())
```

## Specify the size of the n-grams

```python
cv = CountVectorizer(ngram_range=(1, 3))       
ngrams_vectors = cv.fit_transform(df['text'])
print(ngrams_vectors.shape)  
(3088, 70613)                                  
 
pd.DataFrame(ngrams_vectors.toarray(), index = df['text'], columns = cv.get_feature_names_out()).head(10)
```

## Remove Stopwords

```python
cv = CountVectorizer(stop_words='english', max_features=10)  
stopword_filtered = cv.fit_transform(df['text']) 
print(stopword_filtered.shape)  
pd.DataFrame(stopword_filtered.todense(), columns=cv.get_feature_names_out())
```
