# Flight Data Fusion

```python
import pandas as pd
```

### Input Files

```python
input_file_aaa = "../data/2011-12-01-dataaa.csv"
input_file_airtravelcenter = "../data/2011-12-01-dataairtravelcenter.csv"
input_file_dfw = "../data/2011-12-01-datadfw.csv"
```

### Read in the files into one data frame

```python
dfaaa = pd.read_csv(input_file_aaa, delimiter=';', header=0)
dfatc = pd.read_csv(input_file_airtravelcenter, delimiter=';', header=0)
dfdfw = pd.read_csv(input_file_dfw, delimiter=';', header=0)
df = pd.concat([dfaaa, dfatc, dfdfw], ignore_index=True).sort_values(by=['flightno'])
df['depsched'] = pd.to_datetime(df['depsched'])
df['depact'] = pd.to_datetime(df['depact'])
df['depgate'] = df['depgate'].astype(str)   
df['arrivalsched'] = pd.to_datetime(df['arrivalsched'])
df['arrivalact'] = pd.to_datetime(df['arrivalact'])
df['arrivalgat'] = df['arrivalgat'].astype(str)
df
```
