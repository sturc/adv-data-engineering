# Data Quality Monitoring with Great Expectations

```python
import great_expectations as gx
import pandas as pd
```

```python
df = pd.read_csv("../data/sales.csv",delimiter=",")
df
```

## Create an Context for GX

```python
context = gx.get_context()
```

```python
data_source = context.data_sources.add_pandas("pandas")
data_asset = data_source.add_dataframe_asset(name="pd dataframe asset")

batch_definition = data_asset.add_batch_definition_whole_dataframe("batch definition")
batch = batch_definition.get_batch(batch_parameters={"dataframe": df})
```

```python
suite_name = "sales_data_quality_suite"
suite = gx.ExpectationSuite(name=suite_name)
# TODO Define expectations
# see https://docs.greatexpectations.io/docs/guides/expectations/expectation_types/
# https://docs.greatexpectations.io/docs/reference/learn/data_quality_use_cases/dq_use_cases_lp


validation_result = batch.validate(suite)
print(validation_result)
```
